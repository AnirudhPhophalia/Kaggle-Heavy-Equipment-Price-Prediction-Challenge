ligthgbm regressor

```python
import lightgbm as lgb
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_log_error

# Define target and features
X = train_df.drop(['TargetValue', 'TargetValue_log', 'TransactionID'], axis=1)
y = train_df['TargetValue_log']

# Align columns - crucial for consistent feature sets
test_id = test_df['TransactionID']
X_test = test_df.drop('TransactionID', axis=1)

# Get common columns
common_cols = list(set(X.columns) & set(X_test.columns))
X = X[common_cols]
X_test = X_test[common_cols]

# Further align columns to ensure they are in the same order
X_test = X_test[X.columns]

# KFold for cross-validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

rmsle_scores = []
predictions = np.zeros(len(X_test))

for fold, (train_index, val_index) in enumerate(kf.split(X, y)):
    print(f"Fold {fold+1}")
    X_train, X_val = X.iloc[train_index], X.iloc[val_index]
    y_train, y_val = y.iloc[train_index], y.iloc[val_index]

    lgb_params = {
        'objective': 'regression_l1', # MAE objective for robust regression
        'metric': 'rmse',
        'n_estimators': 1000,
        'learning_rate': 0.05,
        'feature_fraction': 0.8,
        'bagging_fraction': 0.8,
        'bagging_freq': 1,
        'verbose': -1, # Suppress verbose output
        'n_jobs': -1,
        'seed': 42 + fold
    }

    model = lgb.LGBMRegressor(**lgb_params)
    model.fit(X_train, y_train, eval_set=[(X_val, y_val)], 
              callbacks=[lgb.early_stopping(100, verbose=False)]) # Fixed line

    val_preds = model.predict(X_val)
    # Ensure predictions are non-negative before inverse log transform
    val_preds[val_preds < 0] = 0
    rmsle = np.sqrt(mean_squared_log_error(np.expm1(y_val), np.expm1(val_preds)))
    rmsle_scores.append(rmsle)
    print(f"RMSLE for Fold {fold+1}: {rmsle:.4f}")

    fold_test_preds = model.predict(X_test)
    # Ensure predictions are non-negative before inverse log transform
    fold_test_preds[fold_test_preds < 0] = 0
    predictions += fold_test_preds / kf.n_splits

print(f"\nAverage RMSLE over {kf.n_splits} folds: {np.mean(rmsle_scores):.4f}")
```

submission file

```python
# Inverse transform the predictions
final_predictions = np.expm1(predictions)

# Ensure predictions are non-negative
final_predictions[final_predictions < 0] = 0

# Create submission DataFrame
submission_df = pd.DataFrame({'TransactionID': test_id, 'TargetValue': final_predictions})

# Save to CSV
submission_df.to_csv('submission.csv', index=False)

print("Submission file 'submission.csv' created successfully.")
display(submission_df.head())
```

### 4. Advanced Feature Engineering: Creating New Features

Now, let's create some new features that might capture more complex relationships within the data.

```python
# Example: Create AssetAge feature
train_df['AssetAge'] = train_df['Year'] - train_df['ManufactureYear']
test_df['AssetAge'] = test_df['Year'] - test_df['ManufactureYear']

print("Created 'AssetAge' feature:")
display(train_df[['ManufactureYear', 'Year', 'AssetAge']].head())
display(test_df[['ManufactureYear', 'Year', 'AssetAge']].head())
```

### 4.1 Frequency Encoding for High-Cardinality IDs

`AssetID` and `ProductConfigID` are high-cardinality integer IDs. Instead of treating them as simple numbers, frequency encoding can provide a more meaningful representation by indicating how common each ID is.

```python
# Frequency Encoding for high-cardinality IDs
for col in ['AssetID', 'ProductConfigID']:
    if col in train_df.columns:
        # Calculate frequencies from the training data
        freq_map = train_df[col].value_counts(normalize=True)

        # Apply frequency encoding to both train and test
        train_df[f'{col}_Freq'] = train_df[col].map(freq_map)
        test_df[f'{col}_Freq'] = test_df[col].map(freq_map)

        # Handle cases where a test value might not be in train_df (assign 0 for unseen)
        test_df[f'{col}_Freq'].fillna(0, inplace=True)

print("\nFrequency encoded 'AssetID' and 'ProductConfigID'.")
display(train_df[['AssetID', 'AssetID_Freq', 'ProductConfigID', 'ProductConfigID_Freq']].head())
display(test_df[['AssetID', 'AssetID_Freq', 'ProductConfigID', 'ProductConfigID_Freq']].head())
```

### 4.2 Interaction Features

Creating interaction terms can help capture synergistic effects between features. Let's combine `AssetAge` and `OperationalHoursMeter`.

```python
# Create interaction feature: AssetAge * OperationalHoursMeter
train_df['Age_Hours_Interaction'] = train_df['AssetAge'] * train_df['OperationalHoursMeter']
test_df['Age_Hours_Interaction'] = test_df['AssetAge'] * test_df['OperationalHoursMeter']

print("\nCreated 'Age_Hours_Interaction' feature.")
display(train_df[['AssetAge', 'OperationalHoursMeter', 'Age_Hours_Interaction']].head())
display(test_df[['AssetAge', 'OperationalHoursMeter', 'Age_Hours_Interaction']].head())
```

### 5. Final Data Preparation for Modeling

Before training new models, we need to clean up the dataframes by dropping redundant columns and ensuring all features are numerical. This includes dropping original `AssetID` and `ProductConfigID` since we have their frequency-encoded versions, and re-doing categorical encoding for any object columns that might have appeared or changed due to new feature engineering.

```python
# Drop original ID columns and TargetValue
# Ensure 'TransactionID' is preserved for submission in test_df

# Columns to drop from both train_df and test_df
drop_cols_from_features = ['AssetID', 'ProductConfigID']

# Drop from train_df
if 'TargetValue' in train_df.columns:
    train_df = train_df.drop(columns=['TargetValue'])
for col in drop_cols_from_features:
    if col in train_df.columns:
        train_df = train_df.drop(columns=[col])

# Drop from test_df
for col in drop_cols_from_features:
    if col in test_df.columns:
        test_df = test_df.drop(columns=[col])

print("Dropped original 'AssetID' and 'ProductConfigID' columns.")
print("Dropped original 'TargetValue' column from train_df.")

# Re-apply Label Encoding for all remaining object columns to ensure consistency
# after new feature engineering. This also handles any new object columns created.
from sklearn.preprocessing import LabelEncoder

# Identify remaining object columns
object_cols_train = train_df.select_dtypes(include='object').columns.tolist()
object_cols_test = test_df.select_dtypes(include='object').columns.tolist()

all_object_cols_final = list(set(object_cols_train + object_cols_test))

print(f"\nRe-encoding the following categorical columns: {all_object_cols_final}")

for col in all_object_cols_final:
    if col in train_df.columns:
        # Fill NA before encoding for categorical columns with mode
        if train_df[col].isnull().any():
            train_df[col] = train_df[col].fillna(train_df[col].mode()[0])

        le = LabelEncoder()
        train_df[col] = le.fit_transform(train_df[col])

    if col in test_df.columns:
        # Fill NA before encoding for categorical columns with mode
        if test_df[col].isnull().any():
            test_df[col] = test_df[col].fillna(test_df[col].mode()[0])

        # Robustly handle unseen labels in test set
        # Create a combined series for fitting the LabelEncoder to ensure consistent mapping
        combined_series = pd.concat([train_df[col], test_df[col]], axis=0).astype(str) # Convert to str to handle mixed types if any
        le_combined = LabelEncoder()
        le_combined.fit(combined_series.unique())

        test_df[col] = le_combined.transform(test_df[col].astype(str))

print("All remaining object columns re-encoded and missing values imputed.")

# Final check for any remaining missing values and object dtypes
print("\nFinal check for missing values in train_df:")
display(train_df.isnull().sum()[train_df.isnull().sum() > 0])
print("\nFinal check for missing values in test_df:")
display(test_df.isnull().sum()[test_df.isnull().sum() > 0])
print("\nFinal check for object dtypes in train_df:")
display(train_df.select_dtypes(include='object').columns)
print("\nFinal check for object dtypes in test_df:")
display(test_df.select_dtypes(include='object').columns)

# Define target and features for modeling
y = train_df['TargetValue_log']
X = train_df.drop(['TargetValue_log', 'TransactionID'], axis=1)

test_id = test_df['TransactionID'] # Store TransactionID for submission
X_test = test_df.drop('TransactionID', axis=1)

# Align columns again after all feature engineering
common_cols = list(set(X.columns) & set(X_test.columns))
X = X[common_cols]
X_test = X_test[common_cols]

# Ensure columns are in the same order
X_test = X_test[X.columns]

print("\nDataframes X, y, X_test prepared for modeling.")
print(f"X shape: {X.shape}")
print(f"X_test shape: {X_test.shape}")
```

### 6. Model Exploration and Comparison

To achieve an RMSLE score below 0.15, we will train and evaluate multiple powerful models, explicitly comparing their performance. We'll start with XGBoost.

#### 6.1 Custom RMSLE Metric

To ensure consistent evaluation across models, let's define a custom RMSLE metric function that we can use for cross-validation.

```python
import numpy as np
from sklearn.metrics import mean_squared_error

def rmsle(y_true, y_pred):
    # Ensure predictions are non-negative
    y_pred[y_pred < 0] = 0
    return np.sqrt(mean_squared_error(np.log1p(y_true), np.log1p(y_pred)))
```

#### 6.2 XGBoost Model Training and Evaluation

Let's train an XGBoost Regressor, leveraging cross-validation for robust evaluation. We'll tune some basic hyperparameters and record its RMSLE score.

```python
import xgboost as xgb
from sklearn.model_selection import KFold

# KFold for cross-validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

xgb_rmsle_scores = []
xgb_predictions = np.zeros(len(X_test))

print("\n--- Training XGBoost Model ---")

for fold, (train_index, val_index) in enumerate(kf.split(X, y)):
    print(f"Fold {fold+1}")
    X_train, X_val = X.iloc[train_index], X.iloc[val_index]
    y_train, y_val = y.iloc[train_index], y.iloc[val_index]

    # XGBoost parameters - tuned slightly for better performance
    # Objective 'reg:squarederror' for RMSE directly on log1p transformed target
    xgb_params = {
        'objective': 'reg:squarederror',
        'eval_metric': 'rmse',
        'n_estimators': 1000, # Will train for this fixed number of estimators
        'learning_rate': 0.05,
        'max_depth': 7,
        'subsample': 0.7,
        'colsample_bytree': 0.7,
        'random_state': 42 + fold,
        'n_jobs': -1,
        'tree_method': 'hist' # For faster training
    }

    model_xgb = xgb.XGBRegressor(**xgb_params)

    # Removed early stopping arguments as they are causing TypeError
    model_xgb.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)

    val_preds_log = model_xgb.predict(X_val)
    # Convert log-transformed predictions back to original scale for RMSLE calculation
    val_preds_original = np.expm1(val_preds_log)

    # Calculate RMSLE using the original y_val (not log-transformed) and original predictions
    # We need the original TargetValue for y_val here, so we inverse transform it from y_val_log
    rmsle_fold = rmsle(np.expm1(y_val), val_preds_original)
    xgb_rmsle_scores.append(rmsle_fold)
    print(f"RMSLE for Fold {fold+1}: {rmsle_fold:.4f}")

    fold_test_preds_log = model_xgb.predict(X_test)
    xgb_predictions += np.expm1(fold_test_preds_log) / kf.n_splits

print(f"\nAverage RMSLE for XGBoost over {kf.n_splits} folds: {np.mean(xgb_rmsle_scores):.4f}")
```

#### 6.3 CatBoost Model Training and Evaluation

Next, let's train a CatBoost Regressor. CatBoost is known for its ability to handle categorical features automatically and its robustness to overfitting. We'll use cross-validation and compare its RMSLE with LightGBM and XGBoost.

```python
import catboost as cb

cat_rmsle_scores = []
cat_predictions = np.zeros(len(X_test))

print("\n--- Training CatBoost Model ---")

for fold, (train_index, val_index) in enumerate(kf.split(X, y)):
    print(f"Fold {fold+1}")
    X_train, X_val = X.iloc[train_index], X.iloc[val_index]
    y_train, y_val = y.iloc[train_index], y.iloc[val_index]

    # CatBoost parameters
    cat_params = {
        'objective': 'RMSE',
        'iterations': 1000,
        'learning_rate': 0.05,
        'depth': 7,
        'l2_leaf_reg': 3,
        'random_seed': 42 + fold,
        'verbose': 0, # Suppress verbose output
        'early_stopping_rounds': 100 # CatBoost handles this directly in fit
    }

    model_cat = cb.CatBoostRegressor(**cat_params)

    # Convert all columns to string type for CatBoost, especially important for 'object' dtype which we already encoded
    # CatBoost can automatically handle categorical features but requires them to be specified
    # Or, as we have already LabelEncoded, they are numerical, so CatBoost will treat them as such.
    # If we had original string categories, we would pass 'cat_features=categorical_features_indices'

    model_cat.fit(X_train, y_train, eval_set=[(X_val, y_val)], early_stopping_rounds=100, verbose=0)

    val_preds_log = model_cat.predict(X_val)
    val_preds_original = np.expm1(val_preds_log)
    val_preds_original[val_preds_original < 0] = 0

    rmsle_fold = rmsle(np.expm1(y_val), val_preds_original)
    cat_rmsle_scores.append(rmsle_fold)
    print(f"RMSLE for Fold {fold+1}: {rmsle_fold:.4f}")

    fold_test_preds_log = model_cat.predict(X_test)
    cat_predictions += np.expm1(fold_test_preds_log) / kf.n_splits

print(f"\nAverage RMSLE for CatBoost over {kf.n_splits} folds: {np.mean(cat_rmsle_scores):.4f}")
```

### 7. Model Comparison and Selection

Let's summarize the performance of all trained models (LightGBM, XGBoost, CatBoost) and select the best one based on the average RMSLE score.

```python
print("\n--- Model Performance Comparison ---")
print(f"LightGBM Average RMSLE: {np.mean(rmsle_scores):.4f}")
print(f"XGBoost Average RMSLE:  {np.mean(xgb_rmsle_scores):.4f}")
print(f"CatBoost Average RMSLE: {np.mean(cat_rmsle_scores):.4f}")

# Select the best model based on average RMSLE
model_names = ['LightGBM', 'XGBoost', 'CatBoost']
avg_rmsle_values = [
    np.mean(rmsle_scores),
    np.mean(xgb_rmsle_scores),
    np.mean(cat_rmsle_scores)
]

best_model_idx = np.argmin(avg_rmsle_values)
best_model_name = model_names[best_model_idx]
best_model_rmsle = avg_rmsle_values[best_model_idx]

print(f"\nThe best performing model is: {best_model_name} with an average RMSLE of {best_model_rmsle:.4f}")

# Assign predictions from the best model to final_predictions for submission
if best_model_name == 'LightGBM':
    final_model_predictions = predictions
elif best_model_name == 'XGBoost':
    final_model_predictions = xgb_predictions
elif best_model_name == 'CatBoost':
    final_model_predictions = cat_predictions

print(f"\nPredictions for submission will be based on the {best_model_name} model.")
```
